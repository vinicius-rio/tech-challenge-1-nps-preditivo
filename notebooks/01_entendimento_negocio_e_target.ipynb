Portanto, o problema não é apenas prever uma nota. O problema é **transformar dados operacionais em decisões práticas para melhorar a experiência do cliente**.

## 6. Por que o NPS é importante para um e-commerce?

O **Net Promoter Score (NPS)** é uma métrica de satisfação e lealdade baseada na probabilidade de um cliente recomendar a empresa, produto ou serviço a outras pessoas.

Em um e-commerce, o NPS é especialmente relevante porque a experiência do cliente depende de múltiplas etapas:

- navegação;
- preço;
- compra;
- pagamento;
- prazo de entrega;
- frete;
- atendimento;
- resolução de problemas;
- pós-venda.

Diferentemente de uma loja física, na qual parte da experiência ocorre presencialmente, no e-commerce a confiança do cliente depende fortemente da capacidade operacional da empresa. Um atraso, uma entrega com múltiplas tentativas ou uma reclamação mal resolvida pode comprometer a percepção de toda a jornada.

### Importância do NPS para o e-commerce

| Motivo | Explicação |
|---|---|
| Mede percepção final da jornada | Resume a experiência do cliente após compra, entrega e atendimento |
| Ajuda a identificar detratores | Permite encontrar clientes com maior risco de insatisfação |
| Apoia priorização operacional | Ajuda a decidir onde atuar primeiro: logística, atendimento, produto ou preço |
| Relaciona satisfação e lealdade | Clientes mais satisfeitos tendem a recomprar e recomendar |
| Facilita comunicação executiva | É uma métrica simples de interpretar e acompanhar |

Assim, o NPS funciona como um indicador estratégico da experiência do cliente, pois conecta percepção, operação e potencial de crescimento.

## 8. Como o NPS impacta recompra?

A recompra é um dos principais efeitos esperados de uma boa experiência do cliente.

Clientes satisfeitos tendem a:

- confiar mais na empresa;
- perceber menor risco em novas compras;
- retornar ao canal de venda;
- comprar com menor necessidade de incentivo promocional;
- manter relacionamento mais duradouro com a marca.

## 9. Como o NPS impacta boca a boca?

O boca a boca é um dos efeitos mais importantes da satisfação do cliente, especialmente em mercados digitais.

Clientes promotores tendem a recomendar a empresa para amigos, familiares ou redes sociais. Já clientes detratores podem compartilhar experiências negativas e afetar a reputação da marca.

Em e-commerce, esse efeito é ainda mais relevante porque a decisão de compra é influenciada por:

- avaliações online;
- comentários em redes sociais;
- reputação da loja;
- relatos de prazo de entrega;
- experiências anteriores de outros consumidores.

Assim, um cliente detrator não representa apenas uma experiência negativa isolada. Ele pode influenciar potenciais clientes e aumentar o custo de aquisição da empresa.

Por outro lado, clientes promotores podem reduzir a dependência de mídia paga, fortalecer a reputação e gerar crescimento orgânico.

## 10. Como o NPS impacta market share em e-commerce?

O market share representa a participação da empresa no mercado. Embora o NPS não determine market share sozinho, ele pode influenciá-lo indiretamente.

Um e-commerce com boa experiência tende a fortalecer:

- retenção de clientes;
- frequência de compra;
- reputação da marca;
- diferenciação competitiva;
- confiança do consumidor;
- recomendação espontânea.

Se os clientes têm experiências positivas e voltam a comprar, a empresa aumenta sua capacidade de capturar valor no mercado. Em contrapartida, experiências ruins podem gerar churn, perda de recorrência e migração para concorrentes.

Portanto, melhorar fatores que impactam o NPS pode contribuir para ganho ou defesa de market share, principalmente em mercados competitivos nos quais o cliente possui baixo custo de troca entre plataformas.

# Etapa 2 — Definição da Target

## 12. Qual variável representa a satisfação do cliente?

A variável que representa a satisfação do cliente é:

```text
nps_score
```

Essa variável contém a nota de NPS atribuída pelo cliente após a experiência de compra, variando de **0 a 10**.

Na lógica tradicional do NPS, os clientes podem ser agrupados em três categorias:

| Faixa da nota | Categoria | Interpretação |
|---|---|---|
| 0 a 6 | Detratores | Clientes insatisfeitos ou com maior risco de falar negativamente da marca |
| 7 a 8 | Neutros | Clientes satisfeitos, mas não necessariamente leais |
| 9 a 10 | Promotores | Clientes altamente satisfeitos e mais propensos a recomendar |

Neste projeto, a variável `nps_score` é a melhor representação disponível da satisfação declarada pelo cliente.

````python
# CÉLULA 6 — CRIAÇÃO DE UMA CLASSIFICAÇÃO CONCEITUAL DO NPS

def classificar_nps(nps_score):
    """
    Classifica a nota de NPS conforme a regra tradicional:
    - 0 a 6: detrator
    - 7 a 8: neutro
    - 9 a 10: promotor
    """
    if nps_score <= 6:
        return "detrator"
    elif nps_score <= 8:
        return "neutro"
    else:
        return "promotor"


df["nps_group"] = df["nps_score"].apply(classificar_nps)

df[["nps_score", "nps_group"]].head()
````

````python
# CÉLULA 7 — DISTRIBUIÇÃO INICIAL DOS GRUPOS DE NPS

df["nps_group"].value_counts(normalize=True).mul(100).round(2)
````

## 13. Por que essa variável foi escolhida?

A variável `nps_score` foi escolhida porque representa diretamente a avaliação feita pelo cliente sobre sua experiência.

Ela é adequada como target porque:

1. **Está alinhada ao problema de negócio**  
   O objetivo do projeto é entender e antecipar a satisfação do cliente. A nota de NPS é a variável que melhor representa essa satisfação na base.

2. **É coletada após a jornada de compra**  
   Como a nota é atribuída depois da experiência, ela reflete o resultado percebido pelo cliente após passar pelas etapas de pedido, entrega e atendimento.

3. **Permite análise quantitativa**  
   Por estar em escala numérica de 0 a 10, pode ser analisada como uma variável contínua em estudos exploratórios ou em modelos de regressão.

4. **Permite análise categórica**  
   A mesma variável pode ser transformada em grupos de negócio, como detratores, neutros e promotores. Isso facilita a comunicação com stakeholders.

5. **É interpretável para áreas de negócio**  
   Gestores conseguem compreender facilmente a diferença entre uma nota baixa, intermediária ou alta, sem necessidade de explicações estatísticas complexas.

Portanto, `nps_score` é uma variável adequada porque conecta a experiência operacional do cliente com uma medida objetiva de satisfação declarada.

## 14. Em que momento da jornada essa informação é coletada?

A variável `nps_score` é coletada **após a experiência de compra**.

Isso significa que o cliente responde à pesquisa depois de passar por etapas como:

1. realização do pedido;
2. pagamento;
3. separação e envio;
4. entrega;
5. eventual contato com atendimento;
6. resolução de problemas, quando houver;
7. avaliação final da experiência.

Esse ponto é importante porque a nota de NPS funciona como uma consequência da jornada. Ou seja, ela deve ser interpretada como um resultado final da experiência do cliente, não como uma causa.

Em termos analíticos, variáveis como atraso na entrega, número de contatos com atendimento, reclamações e tempo de resolução podem ser investigadas como possíveis fatores associados à nota final de NPS.

## 15. Existe algum risco de usar essa variável de forma inadequada?

Sim. Apesar de `nps_score` ser uma boa variável para representar satisfação, existem riscos importantes em seu uso.

### 15.1 Risco de confundir correlação com causalidade

Uma análise pode mostrar que clientes com mais reclamações têm NPS menor. Porém, isso não prova automaticamente que a reclamação causou a nota baixa. A reclamação pode ser consequência de outro problema, como atraso, produto incorreto ou falha de comunicação.

Por isso, os resultados devem ser interpretados como associações, não necessariamente como relações causais.

### 15.2 Risco de vazamento de informação

Se o objetivo futuro for prever o NPS antes da aplicação da pesquisa, não se deve utilizar variáveis que só são conhecidas depois da resposta do cliente ou depois do encerramento completo da jornada.

Esse problema é conhecido como **data leakage**. Ele ocorre quando o modelo usa informações que não estariam disponíveis no momento real da previsão.

### 15.3 Risco de simplificar demais a satisfação

O NPS resume a experiência em uma nota de 0 a 10. Isso é útil para gestão, mas pode esconder nuances importantes.

Dois clientes podem dar a mesma nota por motivos diferentes:

- um por atraso na entrega;
- outro por atendimento ruim;
- outro por preço percebido como injusto;
- outro por problema no produto.

Portanto, a nota deve ser analisada junto com variáveis operacionais e, quando possível, comentários qualitativos.

### 15.4 Risco de viés de resposta

Nem todos os clientes respondem pesquisas de satisfação. Em muitos casos, clientes extremamente satisfeitos ou extremamente insatisfeitos podem ser mais propensos a responder.

Isso pode gerar viés na amostra de respostas e fazer com que o NPS observado não represente perfeitamente toda a base de clientes.

### 15.5 Risco de uso punitivo da métrica

Se a empresa usar o NPS apenas para cobrar áreas ou equipes, pode criar incentivos ruins. O ideal é usar a métrica como instrumento de diagnóstico e melhoria contínua, não apenas como indicador de cobrança.

### 15.6 Risco de ignorar o contexto operacional

Uma nota baixa pode estar associada a diferentes causas. Antes de propor ações, é necessário entender o contexto da jornada. Caso contrário, a empresa pode atacar sintomas em vez de causas reais.

## 16. Possíveis estratégias futuras para a variável alvo

A variável `nps_score` pode ser usada de diferentes formas em etapas futuras do projeto.

### Estratégia 1 — Regressão

Usar `nps_score` como uma variável contínua de 0 a 10.

Nesse caso, o objetivo seria prever a nota esperada de satisfação do cliente.

Exemplo:

```text
Prever se um cliente terá NPS 2.5, 6.8 ou 9.1.
```

Essa abordagem preserva a granularidade da nota, mas pode ser mais difícil de comunicar para áreas de negócio.

### Estratégia 2 — Classificação multiclasse

Transformar `nps_score` em três categorias:

```text
detrator, neutro, promotor
```

Essa abordagem é mais alinhada à lógica tradicional do NPS e facilita a leitura gerencial.

### Estratégia 3 — Classificação binária

Transformar `nps_score` em uma variável binária, por exemplo:

```text
insatisfeito = 1, se nps_score <= 6
insatisfeito = 0, se nps_score > 6
```

Essa estratégia pode ser útil se o principal objetivo da empresa for identificar clientes com risco de insatisfação e priorizar ações preventivas.

Para o contexto do negócio, uma classificação binária de risco de detrator pode ser muito útil operacionalmente, pois permite responder:

> **Quais clientes têm maior probabilidade de terminar a jornada insatisfeitos?**
